```python
import requests

messages = []

def chat_with_llm(prompt):

    messages.append({
        "role": "user",
        "content": prompt
    })

    response = requests.post(
        "http://localhost:11434/api/chat",
        json={
            "model": "llama3.2:3b",
            "messages": messages,
            "stream": False
        }
    )

    result = response.json()

    assistant_reply = result["message"]["content"]

    messages.append({
        "role": "assistant",
        "content": assistant_reply
    })

    return assistant_reply
```

```python
print(chat_with_llm("Where is St. Xavier's College in kolkata?"))
```

```python
print(chat_with_llm("Any other thing you know about it which is in kolkata?"))
```
